### 1.2 RFM Segmentation with KMeans

```python
from src.analytics.rfm import build_customer_segmentation

# Build customer segments using RFM and KMeans
logistics_df, rfm, X_train_scaled, X_test_scaled, kmeans = build_customer_segmentation(logistics_df)

# Descriptive summary
print("=="*20)
print("Base RFM Metrics Summary")
print("=="*20)
print(rfm[['recency','frequency','monetary']].describe().round(2))
```

```python
# Evaluate candidate values of k using Elbow and Silhouette methods
inertia = []
silhouette =[]
for cluster in range(2,12): # Silhouette need at least two cluster.
    kmeans = KMeans(n_clusters = cluster,random_state = 42)
    kmeans.fit(X_train_scaled) 

    # Within-cluster sum of squares (Elbow)
    inertia.append(kmeans.inertia_)

    # Evaluate clustering on the held-out test set
    test_label = kmeans.predict(X_test_scaled) # Predicted lables
    silhouette.append(silhouette_score(X_test_scaled,test_label)) 

# Plot cluster validation metrics
fig,axis = plt.subplots(1,2,figsize=(10,5),sharex=True)

# Elbow Curve
axis[0].plot(range(2,12),inertia,marker='D',color='teal')
axis[0].set_xlabel("Number of Clusters (k)")
axis[0].set_ylabel("Inertia")
axis[0].set_title("Elbow Curve")
axis[0].set_xticks(range(2,12))
axis[0].axvline(x=4,color='r',linestyle='--',alpha=0.5,label="Tight Cluster, k = 4")
axis[0].legend()
axis[0].grid(alpha=0.3)

# Silhouette Score
axis[1].plot(range(2,12),silhouette,marker='D',color='blue')
axis[1].set_xticks(range(2,12))
axis[1].axvline(x=4,color='r',linestyle='--',alpha=0.5,label="Well Seperated Cluster, k = 4")
axis[1].legend()
axis[1].grid(alpha=0.3)
axis[1].set_title("Silhouette Score")
axis[1].set_xlabel("Number of Clusters (k)")
axis[1].set_ylabel("Silhouette")
plt.suptitle("RFM Cluster Validation", fontsize=11, fontweight='bold')
plt.tight_layout()
plt.savefig('../outputs/elbow_silhouette_validation.png')
plt.show()
```

- The elbow curve shows inertia dropping sharply up to k=4, then flattens significantly while the silhouette score reaches peak of 0.365. Therefore, choosing the optimal cluster, k=4, balances the tight cluster with inter-cluster separation to avoid overfitting risk of higher clusters.
